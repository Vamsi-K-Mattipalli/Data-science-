```python
from langchain_groq import ChatGroq
from langchain.agents import Tool, initialize_agent, AgentType
from langchain.tools import DuckDuckGoSearchRun
from langchain_experimental.tools.python.tool import PythonREPLTool
```

```python
# Load Tools
search = DuckDuckGoSearchRun()
python_tool = PythonREPLTool()
```

```python
# Wrap Tools
tools = [
    Tool(
        name="Search",
        func=search.run,
        description="Useful for finding current events or financial data"
    ),
    Tool(
        name="Calculator",
        func=python_tool.run,
        description="Useful for doing math and code-based calculations"
    )
]
```

```python
# LLM and Agent Setup
agent = initialize_agent(
    tools=tools,
    llm=llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True
)
```

```python
# Run
response = agent.invoke("Search the latest INFY quarterly revenues and calculate average growth rate.")
print(response)
```

```python
import yfinance as yf
from langchain.agents import Tool

def get_infys_revenue(_: str) -> str:
    ticker = yf.Ticker("INFY.NS")  # INFY.NS for Infosys on NSE
    try:
        df = ticker.quarterly_income_stmt
        revenue = df.loc["Total Revenue"]
        return revenue.to_string()
    except Exception as e:
        return f"Error fetching revenue: {e}"

infy_tool = Tool(
    name="INFYFinance",
    func=get_infys_revenue,
    description="Gets Infosys quarterly revenue using yfinance"
)
```

```python
from langchain.chat_models import ChatOpenAI
from langchain.agents import initialize_agent, AgentType

groq_api="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"
llm = ChatGroq(
    groq_api_key=groq_api,  # Replace with your actual Groq key
    model_name="llama3-8b-8192",       # You can use "llama3-70b-8192" if available
    temperature=0
)

agent = initialize_agent(
    tools=[infy_tool],
    llm=llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True
)

response = agent.run("Get INFY's last 3 quarterly revenues and calculate the average growth rate.")
print(response)
```
