```python
def add_numbers(x): 
    nums = x.split()                  # Splits the string into a list of words/numbers
    total = sum(map(int, nums))      # Converts each item to int, then sums them
    return str(total)                # Converts the result back to string

add_numbers("10 20 30")
```

```python
from langchain.tools import Tool

calculator_tool = Tool(
    name="Calculator",
    func=add_numbers,
    description="Use this to add numbers. Input format: '3 5'"
)


calculator_tool
```

```python
from langchain_groq import ChatGroq
from langchain.agents import initialize_agent, AgentType
from langchain.tools import Tool

# Initialize LLM using Groq's LLaMA3
llm = ChatGroq(
    groq_api_key=groq_api,  # Replace with your actual key
    model_name="llama3-8b-8192"
)

# Create the agent
agent = initialize_agent(
    tools=[calculator_tool],
    llm=llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True
)

# Run the agent
#response = agent.run("Add these numbers: 15 25 60")
response = agent.invoke("Add these numbers: 15 25 60")

print(response)
```
